```python
campaigns = np.array([
[101, 10000, 800, 80, 5000, 7500],
[102, 15000, 1200, 150, 7000, 12000],
[103, 8000, 400, 30, 3000, 4200],
[104, 20000, 1800, 250, 9000, 18000],
[105, 12000, 600, 90, 6000, 8000],
[106, 18000, 1500, 220, 8000, 16000],
[107, 9000, 300, 20, 2500, 3000],
[108, 25000, 2200, 300, 11000, 24000]
])
print(campaigns)

# campaign_id | Impression | clicks  | conversions  | cost  | revenue

# CTR = Clicks / Impressions × 100
# Conversion Rate = Conversions / Clicks × 100
# ROI = (Revenue - Cost) / Cost × 100
```

```python
# find the highest revenue campaing
revenue= campaigns[::,-1]
print(revenue)
print(revenue.max())
print(revenue.argmax())

# find the highest-ROI campaigm

cost = campaigns[::,-2]
print(cost)

ROI= (revenue-cost)/cost*100
print(ROI)
# ROI = (Revenue - Cost) / Cost × 100
high_roi=np.argmax(ROI)
print(high_roi)

# calculate avg CTR
click= campaigns[::,2]
print(click)

impression= campaigns[::,1]
print(impression)

avg_CTR= click/impression*100
print(avg_CTR)

# calculate avg ROT
print(np.average(ROI))

catreria= np.where(ROI>80,"Excellent",np.where(ROI>40,"Good","Poor"))
print(catreria)

```
