### Numpy: Numpy is the fundamental package for scientific computing in python.
* It is a pthon library that provides a multidimensional array,object,various derived object.

### Numpy array vs Python sequences(list,set.....)/ why numpy
* Numpy array have a fixed size at creation,unlike python list(which can grow dynamically).
* the element in in numpy array are all required to be same data types.
* Multiple advance mathematical calculation


### creating Numpy Arrays

```python
import numpy as np
a=np.array([1,2,3])
print(a)
print(type(a))
```

```python
## 2d array
b= np.array([[2,3,4],[3,4,5]])
print(b)
```

```python
# dtype
np.array([1,2,3],dtype=float)
```

```python
np.array([1,2,3],dtype=bool)
```

```python
np.array([1,2,3],dtype=complex)
```

```python
# np.range
a = np.arange(1,11)
a
```

```python
a = np.arange(1,11,2)
a
```

```python
# with reshape
np.arange(1,11).reshape(5,2)
```

```python
np.arange(1,11).reshape(2,5)
```

```python
np.arange(1,13).reshape(4,3)
```

```python
# np.once and np.zeros :: used to intialised value
np.ones((3,4))
```

```python
np.zeros((3,4))
```

```python
# np.random
np.random.random((3,4))
```

```python
# np.linespace # any two value ko substract krenge to same aayega
np.linspace(-10,10,10)
```

```python
# np.identity
np.identity(3)
```
