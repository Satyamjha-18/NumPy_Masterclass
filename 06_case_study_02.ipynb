```python
import numpy as np
```

```python
# inventory across four stores and five products:
inventory = np.array([
[12, 45, 8, 32, 20],
[5, 60, 15, 25, 10],
[30, 18, 6, 40, 22],
[9, 55, 12, 28, 7]
])

# calculate the total inverntory per store.
tot_inv=(np.sum(inventory,axis=1))
print(tot_inv)

# calculate the total inventory per perduct.
prd_inv=(np.sum(inventory,axis=0))
print(prd_inv)

# Find the store with highest inventory
highest_store= np.argmax(tot_inv)
ans=highest_store+1
print(ans)

# find the product with lowest total invenotory:
low_inv_prd= np.argmin(prd_inv)
ans= low_inv_prd+1
print(ans)

# Extract invenotry velue below 10:
low_inv= inventory[inventory<10]
print(low_inv)
```

```python
# count items that require replenishment.(i.e low inventory prod <10)
n=len(low_inv)
print(n)
# method 2
m = np.sum(inventory<10)
print(m)

# create "reorder"/ "ok" labels.
label = np.where(inventory<10,"reoder","Ok")
print(label)

# Increase inventory below 10 by 20%
inc_inven= np.where(inventory<10,inventory*1.2,inventory)
print(inc_inven)

# round the resulting inventory.
print(np.round(inc_inven),1)

# find the median inventory
print(np.mean(inc_inven))

flat_inv= inc_inven.flatten()
print(flat_inv)

print(np.sort(flat_inv))
np.median(flat_inv)

# Extract the last two product.
print(inc_inven[:,-2])
```

### stacking:

```python
a= np.array([10,20,30])
b= np.array([40,50,60])

result = np.vstack((a,b))
print(result)
```

```python
a= np.array([10,20,30,90])
b= np.array([40,50,60])

result = np.hstack((a,b))   # do not apply v.stack
print(result)
```

use case stacking: 1. jan- feb 
                   2. apr - june

```python
jan_sales = np.array([70, 80, 90])
feb_sales = np.array([65, 75, 85])
mar_sales = np.array([88, 92, 95])

overall_sales = np.vstack((jan_sales,feb_sales,mar_sales))
print(overall_sales)
```

```python
qty_w1 = np.array([70, 80, 90,100,20,90,70]).reshape(-1,1)
qty_w2 = np.array([65, 75, 85,100,20,90,70]).reshape(-1,1)
qty_w3 = np.array([88, 92, 95,100,20,90,70]).reshape(-1,1)
qty_w4=np.array([70, 80, 90,100,20,90,70]).reshape(-1,1)
overall_inventory = np.hstack((qty_w1,qty_w2,qty_w3,qty_w4))
print(overall_inventory)
```
