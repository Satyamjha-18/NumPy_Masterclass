```python
import numpy as np
```

```python
# inventory across four stores and five products:
inventory = np.array([
[12, 45, 8, 32, 20],
[5, 60, 15, 25, 10],
[30, 18, 6, 40, 22],
[9, 55, 12, 28, 7]
])

# calculate the total inverntory per store.
tot_inv=(np.sum(inventory,axis=1))
print(tot_inv)
# calculate the total inventory per perduct.
tot_store=(np.sum(inventory,axis=0))
print(tot_store)
# Find the store with highest inventory
highest_store= np.argmax(tot_inv)
ans=highest_store+1
print(ans)

# find the product with lowest total invenotory:
```

```python
# Calculate total inventory per product
tot_prd = np.sum(inventory, axis=0)
print(tot_prd)

# Find the product with lowest total inventory
low_inv_prd = np.argmin(tot_prd)
print("Product with lowest inventory:", low_inv_prd + 1)
```
