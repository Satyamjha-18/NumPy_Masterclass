```python
# inventory across four stores and five products:
inventory = np.array([
[12, 45, 8, 32, 20],
[5, 60, 15, 25, 10],
[30, 18, 6, 40, 22],
[9, 55, 12, 28, 7]
])

# calculate the total inverntory per store.
tot_inv=(np.sum(inventory,axis=1))
print(tot_inv)

# calculate the total inventory per perduct.
prd_inv=(np.sum(inventory,axis=0))
print(prd_inv)

# Find the store with highest inventory
highest_store= np.argmax(tot_inv)
ans=highest_store+1
print(ans)

# find the product with lowest total invenotory:
low_inv_prd= np.argmin(prd_inv)
ans= low_inv_prd+1
print(ans)

# Extract invenotry velue below 10:
low_inv= inventory[inventory<10]
print(low_inv)
```

```python
# count items that require replenishment.(i.e low inventory prod <10)
n=len(low_inv)
print(n)
# method 2
m = np.sum(inventory<10)
print(m)

# create "reorder"/ "ok" labels.
label = np.where(inventory<10,"reoder","Ok")
print(label)

# Increase inventory below 10 by 20%
inc_inven= np.where(inventory<10,inventory*1.2,inventory)
print(inc_inven)

# round the resulting inventory.
print(np.round(inc_inven),1)

# find the median inventory
print(np.mean(inc_inven))

flat_inv= inc_inven.flatten()
print(flat_inv)

print(np.sort(flat_inv))
np.median(flat_inv)

# Extract the last two product.
print(inc_inven[:,-2])
```

```python
prices = np.array([
820, 835, 810, 850, 875,
860, 890, 910, 895, 930,
945, 920, 960, 975, 990
])

# 1.calculate percentage change between consecutive days(difference).
diff= prices[1:]-prices[:-1]
print(diff)

perc_change=np.round((diff*100)/prices[:-1],2)
print(perc_change)

# 2. Find days where prices increased by more than 2%
high_change_days= perc_change[perc_change>2]
print(high_change_days)

high_change_days= np.where(perc_change>2,"yes","no")
print(high_change_days)

# find maximum, minimum and range.
print(np.max(perc_change))
print(np.min(perc_change))
print(np.max(perc_change)-(np.min(perc_change)))

# Reshape prices into 3x5
new_prices= prices.reshape(3,5)
print(new_prices)

# find avg prices of each row
avg_price= np.mean(new_prices,axis=1)
print(avg_price)

```
