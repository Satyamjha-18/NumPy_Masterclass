### set function
* np.union 1d
* np.intersect 1d
* np.setdiff 1d
* np.setror 1d
* np.in 1d

```python
m= np.array([1,2,3,4,5])
n=np.array([3,4,5,6,7])
print(np.union1d(m,n))
print(np.intersect1d(m,n))
print(np.setdiff1d(m,n))
```
