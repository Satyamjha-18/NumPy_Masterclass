```python
quantity = np.array([2,5,3,4,8,6,10,1,7,5,9,3])

price = np.array([
    1200,800,1500,450,2200,900,
    700,1800,650,1300,500,2500
])

# calculate revenue for every product
rev=quantity*price
print(rev)

# calculate total revenue
print(np.sum(rev))

# find the product index genrating the highest revenue
print(np.argmax(rev))

# find the products whose revenue is above the avg revenue
avg= np.mean(rev)
print(avg)
print(rev[rev>avg])
print(np.where(rev>avg,"yes","no"))

# Apply a 10% discount only to product whose price is greater than 1000.
print(np.where(price>1000,price*0.9,price))

# Round the discount price two decimal prices.
print(np.round(np.where(price>1000,price*0.9,price),2))
# calculate the total quantity sold.
print(np.sum(quantity))
# find the mediun product price
median=np.median(price)
print(median)

# Identify products whose prices is above the median and whose quantity is at least 5.
criteria=(price > median) & (quantity>=5)
print(price[criteria])
```

```python
# A factory collect temp from 20 sensor reading

temp = np.array([
    28.5, 31.2, 29.8, 35.6, 38.1,
40.2, 27.9, 30.5, 42.3, 39.8,
36.4, 33.2, 29.1, 41.7, 44.2,
31.8, 34.5, 37.6, 26.9, 43.1
])

# Find mean ,mediun ,min and max temp
print(temp.mean())
print(np.median(temp))
print(np.min(temp))
print(np.max(temp))

# 2.calculate varience and std deviation.
std=np.std(temp)
print(std)
print(np.square(std))

# 3. Find all temp more than one std away the mean
mean=np.mean(temp)
new_temp= temp[temp>(mean+std)]
print(new_temp)

np.where (temp>(mean+std),"yes","No")

# calculate the absulute difference between every temp and the mean
dev=np.abs(temp-mean)
dev

# Find the five rading having the largest deviation from the mean.
largest_dev= np.sort(dev)
print(largest_dev)

largest_dev[-5:]

# round all temp to nearest whole number.
np.round(temp)
```

```python
patients = np.array([
[101, 120, 85, 98],
[102, 135, 92, 96],
[103, 110, 78, 99],
[104, 150, 88, 94],
[105, 125, 95, 97],
[106, 160, 82, 93],
[107, 115, 90, 98],
[108, 145, 87, 95],
[109, 130, 91, 96],
[110, 155, 80, 92]
])

# Patient id , heart_rate , oxygen layer, recovery_score

# calculate avg heart rate.
print(np.mean(patients[::,1]))

# calaculate medium oxygen level 
print(np.median(patients[::,2]))

# Find patient with heart rate above avg and oxygen below 90.
heart_rate = patients[:,1]
avg_heart_rate= np.mean(heart_rate)
avg_heart_rate

oxy_level= patients[:,2]
mid_oxy=np.median(oxy_level)
mid_oxy
criteria=(heart_rate > avg_heart_rate)& (oxy_level<90)
print(patients[criteria])

# Categorize recovery:>= 97 → Excellent,>= 95 → Stable,Otherwise → Monitor
#print(patients)
recovery=patients[::,-1]
print(recovery)
category=(np.where(recovery>=97,"Excellent",np.where(recovery>=95,"Stable","Monitor")))
print(category)

# count each category
print(np.sum(category=="Excellent"))
print(np.sum(category=="Stable"))
print(np.sum(category=="Monitor"))

# find patient with highest heart rate.
print(heart_rate.max())
print(np.argmax(heart_rate))
patients[np.argmax(heart_rate)]

# Find oxygen level more than 1 std below the mean
std=np.std(patients[::,2])
print(std)
oxy_level_mean= np.mean(patients[:,2])
print(oxy_level)

oxy_level[oxy_level > (oxy_level_mean-std)]
```
