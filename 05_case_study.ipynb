```python
# A teacher has recorded marks of 10 student :
marks = np.array([72,85,92,64,79,88,95,83,70,76])

# Total marks
print(np.sum(marks))
# Averagre marks 
print(np.mean(marks))
print(np.average(marks))
# Highest marks
print(np.max(marks))
# lowest marks 
print(np.min(marks))
# Difference highest and lowest
print(np.max(marks)-np.min(marks))
# Index of the student with highest marks: use np.argmax()
print(np.argmax(marks))
```

```python
prices = np.array([
    499,799,1299,2499,3999,599
])

# prices between 600 and 1000
print(prices>=600)

print(prices <=1000)

print(prices[(prices >=600) & (prices <=1000)])

# prices less than 600 or greater than 1000
print(prices[(prices<=600)| (prices >=1000)])

# prices not greater than 900
print(prices[prices<=900])

print(prices[~ (prices>900)])
```

```python
# An electricity board record the monthly electricity comsumption of a household for one year.
consumption=np.array([
    180,220,195,250,
    280,310,295,270,
    240,210,190,230
])

# 1. Find the total annual consumption.
print("Total consumption:",np.sum(consumption))

# 2. Calculate avg monthly consumption.
print("avg consumption:",np.average(consumption))

# 3.Find the highest and lowest consuption.
print("highest consuption:",np.max(consumption))
print("lowest_consuption:",np.min(consumption))

# Find the month index of highest consumption
print(np.argmax(consumption))

# 4. Extract the first 6 month 
print("First six month:",consumption[0:5])

# 5.Extract the alternate month.
print(consumption[::2])

# 6.Find all month where consumption was greater than annual avg
print(consumption[consumption>np.average(consumption)])

# 7.calculate the percantage increase required if next year consuption is expected to be 8% higher
```

```python
#  The following array contain marks of 5 student in 4 subjects:

marks = np.array([
    [71,85,90,78],
    [65,70,75,80],
    [88,92,95,90],
    [55,60,58,65],
    [78,82,85,88]
])

# 1. shape and size of the array.
print("shape od array:",np.shape(marks))
print("size of arrayy:",np.size(marks))

# 2. marks obtained by the third student.
print(marks[2])

# 3. avg marks of every student row:0 & column :1
stud_avg=print(np.average(marks,axis=1))
stud_avg
# 4. avg marks of every sub
print(np.mean(marks,axis=0))

# 5.student whose avg is above 75
(marks[stud_avg ]>75)

# 6. highest marks obtained in each subject.
print(np.max(marks,axis=0))

# 7. the complete row of the student with the highest avgerage.
ind=np.argmax(stud_avg)
marks[ind]

```

```python
# An online store has the following product prices:

prices = np.array([
    499.50,799.90,1299.75,
    2499.50,3999.90,599.25,
    899.80,1599.40
])
# round prices to the nearest whole number.
print(np.round(prices,1))

# Find the product costing more than 1000.
print(prices[prices>1000])
# calculate a 15% discount only on product costing than 1000
discount= prices[prices>1000]*0.85
discount
# calculate the avg discount price
np.average(discount)

# find most expensice original product
np.max(prices)
```
