```python
# generate 20 random prob values and find their min,max,avg and percentage >0.5

prob=np.random.random(20)
print(prob)
print(np.mean(prob))
print(np.max(prob))
print(np.min(prob))
```

```python
high_prob= prob[prob>0.5]
print(high_prob)

perc=len(high_prob)*100/20
print(perc)
```
